# Seasonal Agriculture Performance Analysis
This notebook analyzes agricultural performance across different seasons using the provided dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = '/content/seasonal_agriculture_performance_dataset.csv'
df = pd.read_csv(file_path)

# Display the first few rows
display(df.head())
```

```python
# Dataset Information
print("Dataset Shape:", df.shape)
print("\nColumn Summary:")
display(df.info())

print("\nMissing Values:")
display(df.isnull().sum())
```

```python
# Data Cleaning: Handling missing values
# Imputing missing values with the median for numeric columns
columns_with_nan = ['Rainfall_mm', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha']
for col in columns_with_nan:
    df[col] = df[col].fillna(df[col].median())

print("Missing values after cleaning:")
print(df[columns_with_nan].isnull().sum())

# Descriptive Statistics
display(df.describe())
```

```python
# Exploratory Data Analysis: Seasonal Profit Analysis
plt.figure(figsize=(10, 6))
sns.boxplot(x='Season', y='Profit_INR', data=df)
plt.title('Profit Distribution by Season')
plt.show()

# Average yield by crop type
avg_yield = df.groupby('Crop')['Yield_Tonnes_Ha'].mean().sort_values(ascending=False)
plt.figure(figsize=(12, 6))
avg_yield.plot(kind='bar', color='skyblue')
plt.title('Average Yield by Crop Type')
plt.ylabel('Yield (Tonnes/Ha)')
plt.xticks(rotation=45)
plt.show()
```

## Advanced Analysis: Environmental Factors and Irrigation Impacts
In this section, we examine correlations between key metrics and the impact of irrigation methods on farm performance.

```python
# Correlation Analysis
plt.figure(figsize=(12, 10))
correlation_matrix = df[['Rainfall_mm', 'Avg_Temperature_C', 'Humidity_pct', 'Soil_Moisture_pct', 'Yield_Tonnes_Ha', 'Profit_INR']].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Key Agriculture Metrics')
plt.show()

# Irrigation Method Impact on Profit
plt.figure(figsize=(10, 6))
sns.barplot(x='Irrigation_Method', y='Profit_INR', data=df)
plt.title('Average Profit by Irrigation Method')
plt.show()
```

## Final Summary and Conclusion
Based on the analysis, we can conclude the seasonal performance and identify areas for optimization.

```python
# Summary Statistics for Conclusion
summary = df.groupby('Season')[['Yield_Tonnes_Ha', 'Profit_INR', 'Water_Used_m3']].mean()
print("Seasonal Performance Summary:")
display(summary)

# Identify most profitable crop-season combination
best_combo = df.groupby(['Season', 'Crop'])['Profit_INR'].mean().idxmax()
best_profit = df.groupby(['Season', 'Crop'])['Profit_INR'].mean().max()

print(f"\nThe most profitable combination is {best_combo} with an average profit of {best_profit:.2f} INR.")
```
